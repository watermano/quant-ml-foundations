# Quant Simulation Task

```python
np.random.seed(42)
```

```python
#Create date
dates= pd.date_range("2024-01-01", periods=1000)

#Simulate returns
returns  = np.random.normal(0.0005, 0.01, 1000)

#Convert to prices
prices = 100*np.exp(np.cumsum(returns))

df= pd.DataFrame({
    "dates": dates,
    "returns": returns,
    "price": prices
})
```

```python
df.head()
```

```python
#Plot Price

plt.plot(df["dates"],df["price"])
plt.title("Simulated Stock Price")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Plot return distributions

plt.hist(df["returns"], bins=50)
plt.title("Return Distribution")
plt.show()
```

```python
#stats
print("Mean return:",df["returns"].mean())
print("Volatility:" ,df["returns"].std())
```
